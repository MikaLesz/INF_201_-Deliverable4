## 1. Temperature class

**Purpose:** Represent a temperature and support Celsius, Fahrenheit and Kelvin.

For this class we show:
- Short description
- Public methods and properties
- A small live example of the functionality

### 1.1 Temperature class overview

**Purpose:** Store a temperature value together with its unit.

**Attributes:**
- `value` – the temperature value
- `unit` – the temperature unit (`"C"`, `"F"` or `"K"`)

**Methods and properties:**
- `convert_to(unit)` – converts the temperature to another unit
- Comparison methods – allow temperatures in different units to be compared
- `__repr__()` / `__str__()` – string representations

```python
print("=== Temperature class overview ===")
print("Public members:")

for name, member in inspect.getmembers(Temperature):
    if name in (
        "__init__", "__repr__", "__str__",
        "convert_to", "to_celsius", "to_fahrenheit", "to_kelvin",
        "__eq__", "__ne__", "__lt__", "__le__", "__gt__", "__ge__"
    ) or not name.startswith("_"):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
temp_c = Temperature(25, "C")
temp_f = Temperature(77, "F")

print(temp_c)
print(temp_f)
print()
print("Attributes of temp_c:", vars(temp_c))
```

### 1.2 Initialization and default unit

**Purpose:** Show how Temperature objects are created.

The class supports:
- A temperature with a specified unit
- A default unit when no unit is supplied
- Unit names written in different letter cases
- Read-only value and unit attributes

```python
print("=== Temperature initialization ===")

default_temp = Temperature(20)
celsius_temp = Temperature(20, "C")
fahrenheit_temp = Temperature(68, "F")
kelvin_temp = Temperature(293.15, "K")

print("Default unit:", default_temp)
print("Celsius:     ", celsius_temp)
print("Fahrenheit:  ", fahrenheit_temp)
print("Kelvin:      ", kelvin_temp)
```

### 1.3 Validation

**Purpose:** Make sure that only valid temperatures and units can be created.

The class checks that:
- The unit is one of `"C"`, `"F"` or `"K"`
- The temperature is a valid number
- The temperature is not below absolute zero

Invalid values should raise a `ValueError`.

```python
print("=== Temperature validation ===")

invalid_examples = [
    ("Invalid unit", lambda: Temperature(20, "X")),
    ("Below absolute zero in Celsius", lambda: Temperature(-274, "C")),
    ("Below absolute zero in Fahrenheit", lambda: Temperature(-460, "F")),
    ("Below absolute zero in Kelvin", lambda: Temperature(-1, "K")),
]

for description, create_temperature in invalid_examples:
    try:
        create_temperature()
    except ValueError as error:
        print(f"{description}: ValueError -> {error}")
```

### 1.4 Temperature conversions

**Purpose:** Demonstrate conversions between Celsius, Fahrenheit and Kelvin.

The same physical temperature can be represented using different units.

**Reference points:**
- `0 °C` = `32 °F` = `273.15 K`
- `100 °C` = `212 °F` = `373.15 K`
- `-40 °C` = `-40 °F` = `233.15 K`

```python
print("=== Temperature conversions ===")

temp = Temperature(25, "C")

print("Original temperature:", temp)
print()
print("--- Using convert_to() ---")
print("25 °C to Fahrenheit:", temp.convert_to("F"))
print("25 °C to Kelvin:    ", temp.convert_to("K"))

temp_f = Temperature(77, "F")
print("77 °F to Celsius:   ", temp_f.convert_to("C"))
```

### 1.5 Round-trip conversion

**Purpose:** Check that converting a temperature to other units and back gives approximately the original value.

Here the temperature is converted from Celsius to Fahrenheit, then to Kelvin, and finally back to Celsius.

```python
print("=== Round-trip conversion ===")

original = Temperature(37, "C")
fahrenheit = original.convert_to("F")
kelvin = fahrenheit.convert_to("K")
back_to_celsius = kelvin.convert_to("C")

print("Original:       ", original)
print("Fahrenheit:     ", fahrenheit)
print("Kelvin:         ", kelvin)
print("Back to Celsius:", back_to_celsius)
print("Same temperature:", original == back_to_celsius)
```

### 1.6 Comparisons

**Purpose:** Compare Temperature objects, including objects that use different units.

The class supports:
- `==` and `!=`
- `<` and `<=`
- `>` and `>=`

The comparison is based on the actual temperature, not the numerical value alone.

```python
print("=== Temperature comparisons ===")

temp_a = Temperature(0, "C")
temp_b = Temperature(32, "F")
temp_c = Temperature(100, "C")

print("temp_a:", temp_a)
print("temp_b:", temp_b)
print("temp_c:", temp_c)
print()
print("temp_a == temp_b:", temp_a == temp_b)
print("temp_a != temp_c:", temp_a != temp_c)
print("temp_a < temp_c: ", temp_a < temp_c)
print("temp_a <= temp_b:", temp_a <= temp_b)
print("temp_c > temp_a: ", temp_c > temp_a)
print("temp_c >= temp_b:", temp_c >= temp_b)
```

### 1.7 Formatting

**Purpose:** Show the difference between `str()` and `repr()`.

- `str()` gives a readable representation for the user.
- `repr()` gives a representation intended to describe the object.

```python
print("=== Temperature formatting ===")

temp = Temperature(21.5, "C")

print("str(temp): ", str(temp))
print("repr(temp):", repr(temp))
```

## 2. Fraction class

**Purpose:** Represent a fraction using a numerator and a denominator.

For this class we show:
- Short description
- Public attributes and methods
- Comparisons
- Arithmetic operations
- Other special methods
- Small live examples

### 2.1 Fraction class overview

**Attributes:**
- `numerator` – the numerator of the fraction
- `denominator` – the denominator of the fraction

**Methods and properties:**
- `__init__()` – creates and reduces a fraction
- `__repr__()` / `__str__()` – string representations
- Comparison methods – compare fractions by their value
- Arithmetic methods – add, subtract, multiply and divide fractions

```python
print("=== Fraction class overview ===")
print("Public members:")

for name, member in inspect.getmembers(Fraction):
    if name in (
        "__init__", "__repr__", "__str__",
        "__eq__", "__ne__", "__lt__", "__le__", "__gt__", "__ge__",
        "__add__", "__sub__", "__mul__", "__truediv__",
        "__neg__", "__abs__", "__pow__", "__float__", "__bool__"
    ) or not name.startswith("_"):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
fraction_a = Fraction(3, 4)
fraction_b = Fraction(1, 2)

print(fraction_a)
print(fraction_b)
print("Attributes of fraction_a:", vars(fraction_a))
```

### 2.2 Initialization and reducing fractions

**Purpose:** Show how Fraction objects are created.

The class supports:
- A numerator and denominator
- A default denominator of `1`
- Reducing fractions to their simplest form
- Rejecting a denominator of zero or a negative denominator

```python
print("=== Fraction initialization ===")

fraction_a = Fraction(3, 4)
fraction_b = Fraction(5)
fraction_c = Fraction(6, 8)

print("Fraction(3, 4):", fraction_a)
print("Fraction(5):   ", fraction_b)
print("Fraction(6, 8):", fraction_c)
print("Numerator:", fraction_c.numerator)
print("Denominator:", fraction_c.denominator)

print("\n--- Invalid denominator ---")
try:
    Fraction(1, 0)
except ValueError as error:
    print("ValueError ->", error)
```

### 2.3 Comparisons

**Purpose:** Compare fractions using their mathematical value.

The class supports all six comparison operators:
- `==` – equal to
- `!=` – not equal to
- `<` – less than
- `<=` – less than or equal to
- `>` – greater than
- `>=` – greater than or equal to

```python
print("=== Fraction comparisons ===")

fraction_a = Fraction(1, 2)
fraction_b = Fraction(2, 4)
fraction_c = Fraction(3, 4)

print("fraction_a:", fraction_a)
print("fraction_b:", fraction_b)
print("fraction_c:", fraction_c)
print()
print("fraction_a == fraction_b:", fraction_a == fraction_b)
print("fraction_a != fraction_c:", fraction_a != fraction_c)
print("fraction_a < fraction_c: ", fraction_a < fraction_c)
print("fraction_a <= fraction_b:", fraction_a <= fraction_b)
print("fraction_c > fraction_a: ", fraction_c > fraction_a)
print("fraction_c >= fraction_b:", fraction_c >= fraction_b)
```

### 2.4 Arithmetic operations

**Purpose:** Demonstrate the four basic arithmetic operations for fractions.

The class supports:
- Addition (`+`)
- Subtraction (`-`)
- Multiplication (`*`)
- Division (`/`)

The results are returned as Fraction objects and reduced to their simplest form.

```python
print("=== Fraction arithmetic ===")

fraction_a = Fraction(1, 2)
fraction_b = Fraction(1, 3)

print("fraction_a:", fraction_a)
print("fraction_b:", fraction_b)
print()
print("Addition:       ", fraction_a + fraction_b)
print("Subtraction:    ", fraction_a - fraction_b)
print("Multiplication: ", fraction_a * fraction_b)
print("Division:       ", fraction_a / fraction_b)

print("\n--- Division by zero ---")
try:
    fraction_a / Fraction(0, 1)
except ZeroDivisionError as error:
    print("ZeroDivisionError ->", error)
```

### 2.5 Other special methods

**Purpose:** Demonstrate additional special methods implemented by Fraction.

Examples:
- `-fraction` – returns the negative fraction
- `abs(fraction)` – returns the absolute value
- `fraction ** 2` – raises the fraction to a power
- `float(fraction)` – converts the fraction to a floating-point number
- `bool(fraction)` – checks whether the fraction is non-zero

```python
print("=== Other Fraction special methods ===")

fraction = Fraction(-2, 3)

print("Original fraction:", fraction)
print("Negative:         ", -fraction)
print("Absolute value:   ", abs(fraction))
print("Power of 2:       ", fraction ** 2)
print("As float:         ", float(fraction))
print("As bool:          ", bool(fraction))
print("Zero as bool:     ", bool(Fraction(0, 1)))
```

### 2.6 Formatting

**Purpose:** Show the string representations of a Fraction object.

- `str()` gives a readable representation.
- `repr()` gives a representation intended to describe the object.

```python
print("=== Fraction formatting ===")

fraction = Fraction(3, 4)

print("str(fraction): ", str(fraction))
print("repr(fraction):", repr(fraction))
```
